# Step 1: Import required libraries

```python

# Step 1: Import required libraries

import nltk
import pandas as pd

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag

from sklearn.feature_extraction.text import (
    TfidfVectorizer,
    CountVectorizer
)

from sklearn.preprocessing import OneHotEncoder

```

```python
# Step 2: Download required NLTK datasets
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
# Step 3: Create a sample text dataset

documents = [
    "The movie was fantastic and the actors performed brilliantly.",
    "The movie was boring and the story was not interesting.",
    "The actors were amazing and the movie was wonderful."
]
```

```python
# Display the dataset
print("Original Dataset:")
for i, doc in enumerate(documents, 1):
    print(f"Document {i}: {doc}")
```

# TASK 1: STOPWORDS REMOVAL

```python

# Stopwords are common words such as 'the', 'is', 'and'
# that usually carry less useful information for analysis.

stop_words = set(stopwords.words('english'))

# Remove stopwords from each document
cleaned_documents = []

for doc in documents:
    tokens = word_tokenize(doc.lower())

    filtered_tokens = [
        word for word in tokens
        if word.isalpha() and word not in stop_words
    ]

    cleaned_documents.append(" ".join(filtered_tokens))

print("\nAfter Stopwords Removal:")
for doc in cleaned_documents:
    print(doc)

```

# TASK 2: TOKENIZATION

```python
# Word Tokenization: Split text into individual words

sample_text = documents[0]

word_tokens = word_tokenize(sample_text)

print("\nWord Tokenization:")
print(word_tokens)


# Sentence Tokenization: Split text into sentences

sentence_tokens = sent_tokenize(
    "The movie was fantastic. The actors performed brilliantly."
)

print("\nSentence Tokenization:")
print(sentence_tokens)
```

# TASK 3: STEMMING AND LEMMATIZATION

```python

# Stemming reduces words to their root-like form.
# Example: playing -> play, studies -> studi

stemmer = PorterStemmer()

stemmed_words = [
    stemmer.stem(word)
    for word in word_tokens
    if word.isalpha()
]

print("\nStemming:")
print(stemmed_words)


# Lemmatization converts words to their dictionary form.
# Example: running -> running (without POS information)
# Example: better -> good (with correct POS information)

lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word.lower())
    for word in word_tokens
    if word.isalpha()
]

print("\nLemmatization:")
print(lemmatized_words)
```

```python
# TASK 4: PART-OF-SPEECH (POS) TAGGING
```

```python

# POS tagging identifies grammatical categories of words.
# Examples: noun (NN), verb (VB), adjective (JJ)

pos_tags = pos_tag(word_tokens)

print("\nPOS Tagging:")
print(pos_tags)

```

# TASK 5: TF-IDF VECTORIZATION

```python

# TF-IDF assigns importance to words based on their
# frequency in a document and rarity across documents.

tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(
    cleaned_documents
)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
)

print("\nTF-IDF Vectorization:")
print(tfidf_df.round(3))
```

# TASK 6: ONE-HOT ENCODING

```python

# One-hot encoding represents each unique token using
# a binary vector.

unique_tokens = sorted(
    set(word for doc in cleaned_documents for word in doc.split())
)

# Create a dataset containing individual tokens
token_df = pd.DataFrame({"Token": unique_tokens})

# Initialize OneHotEncoder
encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown="ignore"
)

# Fit and transform the tokens
one_hot_matrix = encoder.fit_transform(token_df[["Token"]])

one_hot_df = pd.DataFrame(
    one_hot_matrix.astype(int),
    columns=encoder.get_feature_names_out(["Token"])
)

print("\nOne-Hot Encoding:")
print(one_hot_df)

```

# TASK 7: BAG OF WORDS (BoW)

```python

# Bag of Words represents documents using word frequencies.

bow_vectorizer = CountVectorizer()

bow_matrix = bow_vectorizer.fit_transform(
    cleaned_documents
)

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=bow_vectorizer.get_feature_names_out()
)

print("\nBag of Words Representation:")
print(bow_df)

```

# TASK 8: UNIGRAM, BIGRAM AND N-GRAM GENERATION

```python

# Unigram: Single word
# Bigram: Two consecutive words
# Trigram: Three consecutive words
# N-gram: Sequence of n consecutive words

# Unigram generation
unigram_vectorizer = CountVectorizer(ngram_range=(1, 1))

unigram_matrix = unigram_vectorizer.fit_transform(
    cleaned_documents
)

print("\nUnigrams:")
print(unigram_vectorizer.get_feature_names_out())


# Bigram generation
bigram_vectorizer = CountVectorizer(ngram_range=(2, 2))

bigram_matrix = bigram_vectorizer.fit_transform(
    cleaned_documents
)

print("\nBigrams:")
print(bigram_vectorizer.get_feature_names_out())


# Trigram generation
trigram_vectorizer = CountVectorizer(ngram_range=(3, 3))

trigram_matrix = trigram_vectorizer.fit_transform(
    cleaned_documents
)

print("\nTrigrams:")
print(trigram_vectorizer.get_feature_names_out())


# General N-gram generation
# This generates both unigrams, bigrams, and trigrams.

ngram_vectorizer = CountVectorizer(ngram_range=(1, 3))

ngram_matrix = ngram_vectorizer.fit_transform(
    cleaned_documents
)

print("\nUnigrams, Bigrams, and Trigrams:")
print(ngram_vectorizer.get_feature_names_out())


# ============================================================
# END OF ASSIGNMENT
# ============================================================

print("\nAll NLP preprocessing and feature engineering tasks completed successfully!")
```
