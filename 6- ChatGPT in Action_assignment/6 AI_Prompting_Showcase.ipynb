# Connect Gemini API with Python

```python
# Import required libraries
from google import genai
from getpass import getpass

# Enter Gemini API key securely
api_key = getpass("Enter your Gemini API Key: ")

# Create Gemini client
client = genai.Client(api_key=api_key)

# Select Gemini model
MODEL_NAME = "gemini-3.8-flash"

print("Gemini API connected successfully!")
print("Selected Model:", MODEL_NAME)
```

## Create a reusable Gemini function

```python
# Function to send prompts to Gemini

def ask_gemini(prompt):

    response = client.models.generate_content(
        model=MODEL_NAME,
        contents=prompt
    )

    # Return Gemini's actual response
    return response.text
```

## Execute the four assignment tasks

# Task 1 – The Data Extractor (NLP & JSON Formatting)

```python

# Exact prompt
prompt1 = """
You are an expert data extraction assistant.

Read the following text:

"The new 'Starlight' laptop was released on October 26, 2024.
It features a 14-inch OLED screen, weighs 2.5 lbs,
and the base model with 16GB of RAM costs $1,299.
The main complaint from early reviews is its short 6-hour battery life."

Perform the following tasks:

1. Summarize the text in exactly one sentence.
2. Extract the key details into a valid JSON object.

Include these fields:
Product Name, Release Date, Screen Size, Display Type,
Weight, RAM, Price, Battery Life, Main Complaint.

Use valid JSON syntax with double quotes.
Return the summary and JSON separately.
"""

# Display exact prompt
print("EXACT PROMPT:")
print(prompt1)

# Send prompt to Gemini
task1_response = ask_gemini(prompt1)

# Display actual response
print("\nGEMINI RESPONSE:")
print(task1_response)
```

# Task 2 – The Code Generator (Coding)

```python
# TASK 2: CODE GENERATOR

prompt2 = """
Act as an experienced Python programmer.

Write a Python function named calculate_average()
that accepts a list of numbers and returns their average.

Requirements:
1. Accept a list of numerical values.
2. Use sum() and len() to calculate the average.
3. Handle an empty list by returning 0.
4. Include comments explaining the code.
5. Provide an example function call.
6. Display the expected output.

Write clean, beginner-friendly and executable Python code.
"""

# Display exact prompt
print("EXACT PROMPT:")
print(prompt2)

# Send prompt to Gemini
task2_response = ask_gemini(prompt2)

# Display actual response
print("\nGEMINI RESPONSE:")
print(task2_response)
```

# Task 3 – The Code Translator (Python to JavaScript)

```python
# TASK 3: CODE TRANSLATOR

prompt3 = f"""
Act as an expert Python and JavaScript developer.

Convert the following Python code into equivalent JavaScript.

PYTHON CODE GENERATED IN TASK 2:

{task2_response}

Requirements:
1. Convert the complete Python code into JavaScript.
2. Maintain the same functionality and logic.
3. Handle an empty array by returning 0.
4. Use appropriate JavaScript functions and methods.
5. Include comments explaining the code.
6. Include an example of execution.
7. Display the expected output.

Write simple, beginner-friendly JavaScript code.
"""

# Display exact prompt
print("EXACT PROMPT:")
print(prompt3)

# Send prompt to Gemini
task3_response = ask_gemini(prompt3)

# Display actual response
print("\nGEMINI RESPONSE:")
print(task3_response)
```

# Task 4 – The Social Media Manager (Persona & Content Creation)

```python
# TASK 4: SOCIAL MEDIA MANAGER

prompt4 = """
Act as a witty and enthusiastic social media manager.

Write an engaging tweet announcing the new Starlight laptop.

Product details:
- Product Name: Starlight
- Release Date: October 26, 2024
- Display: 14-inch OLED screen
- RAM: 16GB
- Weight: 2.5 lbs
- Price: $1,299
- Battery Life: 6 hours

Instructions:
1. Write an exciting and attention-grabbing tweet.
2. Highlight the OLED display, lightweight design,
   and 16GB RAM.
3. Mention the price of $1,299.
4. Include exactly TWO emojis.
5. Include the hashtag #TechLaunch.
6. Keep it concise and suitable for Twitter/X.
7. Do not include explanations outside the tweet.

Make it creative, witty, and enthusiastic.
"""

# Display exact prompt
print("EXACT PROMPT:")
print(prompt4)

# Send prompt to Gemini
task4_response = ask_gemini(prompt4)

# Display actual response
print("\nGEMINI RESPONSE:")
print(task4_response)
```
