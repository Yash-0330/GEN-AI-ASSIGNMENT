a) Write a program that takes an integer input from the user and prints whether it is even or odd.

```python
number = int(input("Enter an integer: "))

if number % 2 == 0:
    print(number, "is an even number.")
else:
    print(number, "is an odd number.")
```

 b) Implement a program that checks whether a number is posi ve, nega ve, or zero. 

```python
number = float(input("Enter a number: "))

if number > 0:
    print(number, "is a positive number.")
elif number < 0:
    print(number, "is a negative number.")
else:
    print("The number is zero.")
```
