Step 7: Convert sentences into embeddings

```python
# Convert sentences into numerical embeddings
embeddings = model.encode(sentences)

# Display embeddings
print("Sentence Embeddings:")
print(embeddings)

# Display embedding shape
print("\nEmbedding Shape:", embeddings.shape)
```

Step 8: Calculate cosine similarity

```python
# Select the first sentence embedding
first_embedding = embeddings[0].reshape(1, -1)

# Select the remaining sentence embeddings
other_embeddings = embeddings[1:]

# Calculate cosine similarity
similarity_scores = cosine_similarity(
    first_embedding,
    other_embeddings
)[0]

print("Cosine Similarity Scores:\n")

for i in range(1, len(sentences)):
    print(f"Sentence: {sentences[i]}")
    print(f"Similarity Score: {similarity_scores[i-1]:.4f}")
    print()
```

Step 9: Find the most semantically similar sentence

```python
# Find the index of the highest similarity score
most_similar_index = np.argmax(similarity_scores)

# Get the most similar sentence
most_similar_sentence = sentences[most_similar_index + 1]

# Get the highest similarity score
highest_score = similarity_scores[most_similar_index]

print("Original Sentence:")
print(sentences[0])

print("\nMost Semantically Similar Sentence:")
print(most_similar_sentence)

print("\nHighest Similarity Score:")
print(round(highest_score, 4))
```

Step 10: Complete assignment code

```python
# Assignment: Tokenization and Semantic Similarity

# Step 1: Import libraries
import nltk
import numpy as np

from nltk.tokenize import word_tokenize
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# Step 2: Download NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')

# -----------------------------------------
# PART 1: TOKENIZATION
# -----------------------------------------

# Given sentence
text = "Prompt engineering is the art of guiding AI models effectively."

# Perform tokenization
tokens = word_tokenize(text)

print("========== TOKENIZATION ==========")

print("\nOriginal Sentence:")
print(text)

print("\nTokens:")
print(tokens)

print("\nTotal Tokens:", len(tokens))


# -----------------------------------------
# PART 2: SEMANTIC SIMILARITY
# -----------------------------------------

# Given list of sentences
sentences = [
    "The king is the ruler of the country.",
    "The queen governs her people with wisdom.",
    "I love a good pepperoni pizza.",
    "Monarchy is a form of government."
]

print("\n========== SEMANTIC SIMILARITY ==========")

# Load pretrained Sentence Transformer model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Convert sentences into embeddings
embeddings = model.encode(sentences)

print("\nEmbeddings Shape:", embeddings.shape)

# Get embedding of the first sentence
first_embedding = embeddings[0].reshape(1, -1)

# Get embeddings of the remaining sentences
other_embeddings = embeddings[1:]

# Calculate cosine similarity
similarity_scores = cosine_similarity(
    first_embedding,
    other_embeddings
)[0]

# Display similarity scores
print("\n========== SIMILARITY SCORES ==========")

for i in range(1, len(sentences)):
    print("\nSentence:", sentences[i])
    print("Similarity Score:", round(similarity_scores[i-1], 4))

# Find the most similar sentence
most_similar_index = np.argmax(similarity_scores)

most_similar_sentence = sentences[most_similar_index + 1]

highest_score = similarity_scores[most_similar_index]

# Display final result
print("\n========== FINAL RESULT ==========")

print("\nOriginal Sentence:")
print(sentences[0])

print("\nMost Similar Sentence:")
print(most_similar_sentence)

print("\nSimilarity Score:")
print(round(highest_score, 4))

print("\nAssignment Completed Successfully!")
```
