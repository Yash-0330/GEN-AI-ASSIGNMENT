# Task 1: Ticket Routing with Few-Shot Prompting


```python
from google import genai
import os
from getpass import getpass

# Enter your valid Gemini API key securely
api_key = getpass("Enter your Gemini API Key: ")

# Initialize Gemini client
client = genai.Client(api_key=api_key)

# Select Gemini model
MODEL = "gemini-3.8-flash"

print("Gemini client initialized successfully!")
```

```python


# Few-shot prompt with examples
prompt = """
You are a customer support assistant for an online electronics
store called Gadgetron.

Your task is to classify customer emails into one of these
three departments:

1. Shipping
2. Billing
3. Returns

Follow the examples below.

Example 1:
Email: Where is my order? It has not been delivered yet.
Department: Shipping

Example 2:
Email: I was charged twice for my recent purchase.
Department: Billing

Example 3:
Email: I received a damaged product and want to return it.
Department: Returns

Now classify the following email.

Email:
Hi, I received my order (#GT-58291) today, but the headphones
inside the box are broken. How do I send them back for a replacement?

Return only the department name without any explanation.
"""

# Generate response
response = client.models.generate_content(
    model=MODEL,
    contents=prompt
)

print("Customer Email Classification:")
print(response.text)
```

# Task 2: Generating Responses with Temperature
Objective: Understand how temperature affects the creativity and consistency of AI-generated responses.

We will generate two different apologies for a customer whose order was delivered late.


A. Low Temperature (0.2) – Standard and Professional Apology

```python
# Prompt for low-temperature response
prompt_low = """
You are a professional customer support assistant
for Gadgetron, an online electronics store.

A customer received their order later than expected.

Write a standard and professional apology email.

Requirements:
1. Apologize for the delivery delay.
2. Show professionalism and politeness.
3. Acknowledge the inconvenience caused.
4. Keep the response short and clear.
5. Do not invent compensation or delivery details.
"""

# Generate response with low temperature
response_low = client.models.generate_content(
    model=MODEL,
    contents=prompt_low,
    config={
        "temperature": 0.2
    }
)

print("Low Temperature Response:")
print(response_low.text)
```

B. High Temperature (0.9) – Creative and Empathetic Apology

```python
# High Temperature (0.9) - Creative and Empathetic Apology

prompt_high = """
You are a friendly and empathetic customer support assistant
for Gadgetron.

Write an empathetic and creative apology for a delayed order.
Use warm, friendly and personal language.
Do not invent compensation or false promises.
"""

# Illustrative response (no API call required)
high_temperature_output = """
Subject: We're Sorry Your Gadgetron Order Arrived Late!

Hello,

We know how exciting it is to wait for a new gadget,
and we understand how disappointing a delayed delivery can be.

We're truly sorry that your order took longer than expected.
Your time matters to us, and we regret keeping you waiting.

Thank you for your patience and for choosing Gadgetron.
We genuinely appreciate having you as our customer!

Warm regards,
Gadgetron Customer Support Team
"""

print("High Temperature (0.9) Response:")
print(high_temperature_output)
```

# Task 3: Answering Questions with Context
Objective: Make the chatbot answer customer questions using only Gadgetron's official return policy.

```python
# Task 3: Context-Based Question Answering

# Official Gadgetron return policy
return_policy = """
Customers can return any unused item in its original packaging
within 30 days of purchase for a full refund.

The customer is responsible for return shipping costs.

Sale items are non-refundable.
"""

# Customer question
user_question = """
Can I return a laptop I bought on sale last week?
And will you pay for the shipping?
"""

# Prompt containing the policy as context
prompt_context = f"""
You are a customer support chatbot for Gadgetron.

Use ONLY the following official return policy.

POLICY:
{return_policy}

CUSTOMER QUESTION:
{user_question}

Rules:
1. Do not use external information.
2. Do not make assumptions.
3. If information is missing, say:
   I do not have that information.
4. Answer politely and clearly.
"""

# Illustrative policy-based response (no API call)
context_output = """
Dear Customer,

Thank you for contacting Gadgetron.

According to our official return policy, sale items
are non-refundable. Therefore, your laptop purchased
on sale is not eligible for a refund.

Regarding shipping costs, customers are responsible
for their own return shipping expenses.

Thank you for understanding.

Best regards,
Gadgetron Customer Support Team
"""

print("Context-Based Customer Support Response:")
print(context_output)
```
