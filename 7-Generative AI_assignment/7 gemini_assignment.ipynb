# Connect Gemini with Python

```python

from google import genai
from getpass import getpass

# Securely enter your Gemini API key
API_KEY = getpass("Enter your Gemini API key: ").strip()

# Connect to Gemini
client = genai.Client(api_key=API_KEY)

# Select Gemini model
MODEL = "gemini-3.8-flash"

print("Gemini client initialized successfully!")
```

# Task 1: The Explainer
Objective: Ask Gemini to explain the main difference between GAN and VAE.

```python
prompt = """
Explain the main difference between Generative Adversarial Networks
(GANs) and Variational Autoencoders (VAEs).

Explain in simple language suitable for a beginner.
Include:
1. What is GAN?
2. What is VAE?
3. How do they work?
4. Main differences between GAN and VAE.
5. One real-world application of each.
"""

response = client.models.generate_content(
    model=MODEL,
    contents=prompt
)

print(response.text)
```

# Task 2: The Poet
Objective: Ask Gemini to write a short four-line poem about artificial intelligence.

```python
prompt = """
Write a creative four-line poem about Artificial Intelligence.

Requirements:
1. Exactly four lines.
2. Use simple English.
3. Include creativity, technology, and the future.
4. Make the poem meaningful and inspiring.
"""

response = client.models.generate_content(
    model=MODEL,
    contents=prompt
)

print(response.text)
```

# Task 3: The Coder
Objective: Ask Gemini to write a simple Python function that accepts a name and prints a personalized greeting.

```python
prompt = """
Write a simple Python function that takes a person's name as input
and prints a personalized greeting.

Requirements:
1. Create a function called greet_user().
2. Accept the name as a parameter.
3. Print Hello followed by the person's name.
4. Include comments in the code.
5. Show an example of calling the function.
"""

response = client.models.generate_content(
    model=MODEL,
    contents=prompt
)

print(response.text)
```

```python
# Function to greet a user
def greet_user(name):
    # Print a personalized greeting
    print("Hello, " + name + "! Welcome to Python programming.")

# Calling the function
greet_user("Yash")
```
